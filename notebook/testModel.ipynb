```python
#import necessary package
import torch
from torchvision import transforms
from PIL import Image
```

```python
# Define preprocessing pipeline
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load and preprocess image
img = Image.open("test_fruit_cherry2.jpg")   # replace with your image path
img = transform(img).unsqueeze(0)    # add batch dimension

# Load model
modelTest = torch.load("fruit_mobilenetv2.pth")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
modelTest = modelTest.to(device)
img = img.to(device)

with torch.no_grad():
    outputs = modelTest(img)
    _, predicted = torch.max(outputs, 1)

class_names = ["apple", "Cherry", "Tomatoe"]
print("Predicted fruit:", class_names[predicted.item()])


```
